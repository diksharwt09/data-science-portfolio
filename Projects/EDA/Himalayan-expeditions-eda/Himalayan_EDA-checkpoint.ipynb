# Himalayan Expeditions — Exploratory Data Analysis

Analyzing decades of Himalayan mountaineering expedition data to understand how success rates, safety, team composition, and climbing nations have 
evolved — and to test whether supplemental oxygen genuinely improves outcomes.

**Dataset:** The Himalayan Database — expedition and peak records for Nepal Himalaya climbs.

**Files used:** `exped.csv` (expedition-level records), `peaks.csv`(peak names and heights, merged in for readability).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
exped = pd.read_csv("exped.csv", low_memory = False)
exped.head()
```

```python
exped.isnull().sum().sort_values(ascending=False).head(20)
```

```python
exped.info()
```

```python
key_cols = ['year', 'season', 'nation', 'success1', 'totmembers', 'smtmembers', 'mdeaths', 'tothired', 'hdeaths', 'o2used']
exped[key_cols].isnull().sum()
```

```python
# Success rate by decade
exped['decade'] = (exped['year'] // 10) * 10
success_by_decade = exped.groupby('decade')['success1'].mean() * 100
success_by_decade
```

```python
#Death rate by decade
exped['death_rate'] = exped['mdeaths'] / exped['totmembers']
death_by_decade = exped.groupby('decade')['death_rate'].mean() * 100
death_by_decade
```

```python
# Visual of Success and death rate by decade

fig, axes = plt.subplots(1, 2, figsize=(14,5))

success_by_decade.plot(kind='line', marker='o', ax=axes[0], color='green')
axes[0].set_title('Success Rate by Decade')
axes[0].set_ylabel('Success Rate (%)')

death_by_decade.plot(kind='line', marker='o', ax=axes[1], color='red')
axes[1].set_title('Death Rate by Decade')
axes[1].set_ylabel('Death Rate (%)')

plt.tight_layout()
plt.show()
```

```python
exped.groupby('decade')['expid'].count()
```

```python
exped_modern = exped[exped['year'] >= 1950].copy()

success_by_decade2 = exped_modern.groupby('decade')['success1'].mean() * 100
death_by_decade2 = exped_modern.groupby('decade')['death_rate'].mean() * 100
```

**Note:** Decades before 1950 are excluded from trend analysis due to very small sample sizes (fewer than 35 expeditions per decade before 1950, compared to hundreds-to-thousands per decade afterward). Rates calculated from such small samples are highly sensitive to individual events and not representative of broader trends.

```python
#The 1970 gap

exped_modern[exped_modern['decade'] == 1970][['year','totmembers','mdeaths','death_rate']].head(10)
```

```python
exped_modern = exped_modern[exped_modern['totmembers'] > 0].copy()
exped_modern['death_rate'] = exped_modern['mdeaths'] / exped_modern['totmembers']

death_by_decade2 = exped_modern.groupby('decade')['death_rate'].mean() * 100
death_by_decade2
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14,5))

success_by_decade2.plot(kind='line', marker='o', ax=axes[0], color='green')
axes[0].set_title('Success Rate by Decade (1950+)')
axes[0].set_ylabel('Success Rate (%)')

death_by_decade2.plot(kind='line', marker='o', ax=axes[1], color='red')
axes[1].set_title('Death Rate by Decade (1950+)')
axes[1].set_ylabel('Death Rate (%)')

plt.tight_layout()
plt.show()
```

```python
# Success rate with vs without oxygen 
oxygen_success = exped_modern.groupby('o2used')['success1'].mean() * 100
oxygen_success
```

```python
#Death rate with vs. without oxygen
oxygen_death = exped_modern.groupby('o2used')['death_rate'].mean() * 100
oxygen_death
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12,5))

oxygen_success.plot(kind='bar', ax=axes[0], color=['salmon','skyblue'])
axes[0].set_title('Success Rate: Oxygen Used vs Not')
axes[0].set_xticklabels(['No O2', 'O2 Used'], rotation=0)
axes[0].set_ylabel('Success Rate (%)')

oxygen_death.plot(kind='bar', ax=axes[1], color=['salmon','skyblue'])
axes[1].set_title('Death Rate: Oxygen Used vs Not')
axes[1].set_xticklabels(['No O2', 'O2 Used'], rotation=0)
axes[1].set_ylabel('Death Rate (%)')

plt.tight_layout()
plt.show()
```

**Interpretation caution:** While oxygen use is strongly associated with higher success rates, death rates are nearly identical between the two groups. This does *not* mean oxygen fails to reduce risk — oxygen is used disproportionately on the highest, most dangerous peaks (e.g., Everest), which confounds a direct comparison. A fairer test would control for peak height or  restrict comparison to a single mountain.

```python
#Oxygen comparison, Everest only
everest = exped_modern[exped_modern['peakid'] == 'EVER']

everest_success = everest.groupby('o2used')['success1'].mean() * 100
everest_death = everest.groupby('o2used')['death_rate'].mean() * 100

print("Everest Success Rate by O2 use:")
print(everest_success)
print("\nEverest Death Rate by O2 use:")
print(everest_death)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12,5))

everest_success.plot(kind='bar', ax=axes[0], color=['salmon','skyblue'])
axes[0].set_title('Everest Success Rate: O2 vs No O2')
axes[0].set_xticklabels(['No O2', 'O2 Used'], rotation=0)
axes[0].set_ylabel('Success Rate (%)')

everest_death.plot(kind='bar', ax=axes[1], color=['salmon','skyblue'])
axes[1].set_title('Everest Death Rate: O2 vs No O2')
axes[1].set_xticklabels(['No O2', 'O2 Used'], rotation=0)
axes[1].set_ylabel('Death Rate (%)')

plt.tight_layout()
plt.show()
```

**Controlling for peak difficulty (Everest only):** Once mountain is held constant, oxygen's effect becomes much clearer — non-oxygen climbers on Everest succeed only 6% of the time (vs 80% with oxygen) and face a higher death rate (2.5% vs 1.5%). This confirms the earlier full-dataset comparison was masking oxygen's real safety benefit, since non-oxygen climbs are concentrated on the most extreme peaks.

```python
#Top nationalities by expedition count
top_nations = exped_modern['nation'].value_counts().head(10)
top_nations
```

```python
plt.figure(figsize=(10,5))
top_nations.plot(kind='barh', color='teal')
plt.title('Top 10 Nations by Number of Expeditions')
plt.xlabel('Number of Expeditions')
plt.gca().invert_yaxis()
plt.show()
```

```python
#Team size trend over time 
team_size_by_decade = exped_modern.groupby('decade')['totmembers'].mean()
team_size_by_decade
```

```python
plt.figure(figsize=(8,5))
team_size_by_decade.plot(kind='line', marker='o', color='purple')
plt.title('Average Team Size by Decade')
plt.ylabel('Average Members per Expedition')
plt.xlabel('Decade')
plt.show()
```

```python
numeric_cols = ['year','totmembers','smtmembers','mdeaths','tothired','hdeaths','totdays','death_rate']
corr = exped_modern[numeric_cols].corr()

plt.figure(figsize=(9,7))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap: Expedition Variables')
plt.show()
```

**Correlation takeaways:** Expedition size metrics (total members, summit members, hired staff) move together, as expected. Notably, `year` shows a mild negative correlation with team size and trip duration — consistent with the shift toward smaller, faster, modern-style expeditions seen in the team size chart. No single variable strongly predicts death rate on its own, suggesting safety improvements are driven by a combination of factors not fully captured in this dataset (gear, forecasting, experience).

```python
peaks = pd.read_csv('peaks.csv', low_memory=False)
peaks.head()
```

```python
peaks_slim = peaks[['peakid', 'pkname', 'heightm']]
exped_named = exped_modern.merge(peaks_slim, on='peakid', how='left')
exped_named[['peakid','pkname','heightm']].head()
```

```python
peak_stats = exped_named.groupby(['peakid','pkname']).agg(
    total_expeditions=('expid','count'),
    total_deaths=('mdeaths','sum'),
    avg_death_rate=('death_rate','mean')
).reset_index()

peak_stats_filtered = peak_stats[peak_stats['total_expeditions'] >= 20]
deadliest = peak_stats_filtered.sort_values('avg_death_rate', ascending=False).head(10)
deadliest
```

```python
plt.figure(figsize=(10,6))
plt.barh(deadliest['pkname'], deadliest['avg_death_rate']*100, color='darkred')
plt.xlabel('Average Death Rate (%)')
plt.title('Top 10 Deadliest Peaks (min. 20 expeditions)')
plt.gca().invert_yaxis()
plt.show()
```

##  Summary of Key Findings

**1. Climbing has gotten dramatically safer and more successful since 1950.**
Success rates rose from ~34% in the 1950s to over 70% in the 2020s, while death rates fell from a 1970 peak of ~2.7% to under 1% today — driven by 
better equipment, weather forecasting, and route knowledge over the decades. 
*(Pre-1950 data was excluded from this trend due to very small sample sizes.)*

**2. Oxygen use has a real, measurable safety and success benefit — but only 
visible once you control for peak difficulty.**
Across the full dataset, oxygen users had far higher success rates (80% vs 45%) but nearly identical death rates to non-oxygen climbers — a misleading result caused by oxygen being used disproportionately on the most extreme peaks. Isolating Everest alone revealed the true effect: non-oxygen climbers succeed only 6% of the time (vs 80% with oxygen) and face a notably higher death rate (2.5% vs 1.5%).

**3. Expedition teams have shrunk significantly since the 1970s.**
Average team size peaked at ~10 members in the 1970s ("siege-style" expeditions) and declined to ~5 by the 2010s, reflecting a broader shift toward smaller, lightweight climbing teams.

**4. The USA, Japan, and the UK have sent the most expeditions historically**, 
consistent with the sport's early development in Western and East Asian mountaineering communities.

**5. Annapurna I and Dhaulagiri I rank among the deadliest peaks** 
(minimum 20 expeditions), consistent with their well-known reputations in real-world mountaineering — a good sanity check that this analysis reflects genuine patterns rather than data artifacts.

**A note on data quality:** several data quality issues were identified and handled during this analysis, including small-sample distortion in early decades, a confounding variable in the oxygen comparison, and low-sample peaks producing misleadingly extreme death rates. Addressing these was as important to the analysis as the visualizations themselves.
